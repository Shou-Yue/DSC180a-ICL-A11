```python
import os
import math
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
from transformers import GPT2Config, GPT2Model
```

```python
OUT_DIR = "models/overparameterized_linear_regression"

# model settings
N_DIMS = 20
N_POSITIONS = 101  # this just has to be > the max points in the curriculum
N_EMBD = 256
N_ATT_LAYERS = 12
N_ATT_HEADS = 8

# training settings
BATCH_SIZE = 64
LEARNING_RATE = 1e-4
TRAIN_STEPS = 500_001
SAVE_EVERY_STEPS = 1_000
KEEP_EVERY_STEPS = 100_000

# curriculum settings
CURR_DIMS_START = 20
CURR_DIMS_END = 20
CURR_DIMS_INC = 0
CURR_POINTS_START = 5
CURR_POINTS_END = 15
CURR_POINTS_INC = 1
CURR_INTERVAL = 2_000  # how often to update dims/points

# linear regression setting
SPARSITY = None # none if standard linear regression, number of nonzero weights for sparse
```

```python
def mean_squared_error(pred, target):
    return ((pred - target) ** 2).mean()

# [todo turn start, end, inc into a dataclass]
class Curriculum:
    def __init__(
        self,
        dims_start,
        dims_end,
        dims_inc,
        points_start,
        points_end,
        points_inc,
        interval,
    ):
        self.dims_trunc = dims_start
        self.points = points_start

        self.dims_start = dims_start
        self.dims_end = dims_end
        self.dims_inc = dims_inc

        self.points_start = points_start
        self.points_end = points_end
        self.points_inc = points_inc

        self.interval = interval
        self.step = 0

    def update(self):
        self.step += 1
        if self.step % self.interval == 0:
            self.dims_trunc = min(self.dims_trunc + self.dims_inc, self.dims_end)
            self.points = min(self.points + self.points_inc, self.points_end)
```

```python
class TransformerModel(nn.Module):
    def __init__(self, n_dims, n_positions, n_embd = 256, n_layer = 12, n_head = 8):
        super().__init__()
        config = GPT2Config(
            n_positions = 2 * n_positions,
            n_embd = n_embd,
            n_layer = n_layer,
            n_head = n_head,
            resid_pdrop = 0.0,
            embd_pdrop = 0.0,
            attn_pdrop = 0.0,
            use_cache = False,
        )
        self.name = f"gpt2_embd={n_embd}_layer={n_layer}_head={n_head}"

        self.n_positions = n_positions
        self.n_dims = n_dims

        self.read_in = nn.Linear(n_dims, n_embd)
        self.backbone = GPT2Model(config)
        self.read_out = nn.Linear(n_embd, 1)

    @staticmethod
    def _combine(xs_b, ys_b):
        """
        Interleave x's and y's into a single sequence.

        xs_b: [B, T, D]
        ys_b: [B, T]

        Returns: [B, 2T, D] where positions are x_1, y_1, x_2, y_2, ...
        """
        bsize, points, dim = xs_b.shape

        ys_b_wide = torch.cat(
            (
                ys_b.view(bsize, points, 1),
                torch.zeros(bsize, points, dim - 1, device = ys_b.device),
            ),
            dim = 2,
        )

        # zs[:, 0] = x_1, zs[:, 1] = y_1, zs[:, 2] = x_2, zs[:, 3] = y_2, ...
        zs = torch.stack((xs_b, ys_b_wide), dim = 2)
        zs = zs.view(bsize, 2 * points, dim)
        return zs

    def forward(self, xs, ys):
        """
        xs: [B, T, D]
        ys: [B, T]

        Returns predictions for all y positions: [B, T]
        """
        zs = self._combine(xs, ys)
        embeds = self.read_in(zs)
        output = self.backbone(inputs_embeds = embeds).last_hidden_state
        prediction = self.read_out(output)          # [B, 2T, 1]
        prediction = prediction[:, ::2, 0]          # keep only x positions -> [B, T]
        return prediction
```

```python
def sample_gaussian_xs(batch_size, n_points, n_dims, n_dims_trunc, device):
    """
    Sample Gaussian inputs and optionally truncate to a lower-dimensional
    subspace by zeroing out the last coordinates.
    """
    xs = torch.randn(batch_size, n_points, n_dims, device = device)
    if n_dims_trunc is not None and n_dims_trunc < n_dims:
        xs[:, :, n_dims_trunc:] = 0.0
    return xs


def sample_linear_regression_weights(batch_size, n_dims, device, sparsity_s = None):
    """
    Sample one weight vector per batch element, w ~ N(0, I).
    """
    w = torch.randn(batch_size, n_dims, 1, device = device)

    if sparsity_s is None or sparsity_s >= n_dims:
        return w

    for b in range(batch_size):
        idx = torch.randperm(n_dims, device = device)[:sparsity_s]
        mask = torch.zeros(n_dims, 1, device = device, dtype = torch.bool)
        mask[idx] = True
        w[b, ~mask] = 0.0

    return w


def evaluate_linear_regression(xs, w, scale = 1.0):
    """
    xs: [B, T, D]
    w: [B, D, 1]

    Returns ys: [B, T]
    """
    ys = scale * (xs @ w)   # [B, T, 1]
    return ys[:, :, 0]
```

```python
def train(model):
    device = next(model.parameters()).device

    os.makedirs(OUT_DIR, exist_ok = True)
    state_path = os.path.join(OUT_DIR, "state.pt")

    optimizer = torch.optim.Adam(model.parameters(), lr = LEARNING_RATE)

    curriculum = Curriculum(
        dims_start = CURR_DIMS_START,
        dims_end = CURR_DIMS_END,
        dims_inc = CURR_DIMS_INC,
        points_start = CURR_POINTS_START,
        points_end = CURR_POINTS_END,
        points_inc = CURR_POINTS_INC,
        interval = CURR_INTERVAL,
    )

    # Resume if state exists
    starting_step = 0
    if os.path.exists(state_path):
        state = torch.load(state_path, weights_only = True)
        model.load_state_dict(state["model_state_dict"])
        optimizer.load_state_dict(state["optimizer_state_dict"])
        starting_step = state["train_step"]

        # Fast forward curriculum
        for _ in range(starting_step + 1):
            curriculum.update()

    ema_loss = None
    alpha = 0.01

    pbar = tqdm(range(starting_step, TRAIN_STEPS))
    for step in pbar:
        n_points = curriculum.points
        n_dims_trunc = curriculum.dims_trunc

        # Sample data
        xs = sample_gaussian_xs(
            batch_size = BATCH_SIZE,
            n_points = n_points,
            n_dims = model.n_dims,
            n_dims_trunc = n_dims_trunc,
            device = device,
        )
        w = sample_linear_regression_weights(
            batch_size = BATCH_SIZE,
            n_dims = model.n_dims,
            device = device,
            sparsity_s = SPARSITY
        )
        ys = evaluate_linear_regression(xs, w)  # [B, T]

        # Training step
        model.train()
        optimizer.zero_grad()
        preds = model(xs, ys)
        loss = mean_squared_error(preds, ys)
        loss.backward()
        optimizer.step()

        loss_val = loss.detach().item()
        if ema_loss is None:
            ema_loss = loss_val
        else:
            ema_loss = alpha * loss_val + (1.0 - alpha) * ema_loss

        curriculum.update()

        desc_dict = {
            "loss": round(loss_val, 4),
            "ema_loss": round(ema_loss, 4),
            "d": n_dims_trunc,
            "n": n_points,
        }
        pbar.set_description(str(desc_dict))

        # Save training state (for resume)
        if step % SAVE_EVERY_STEPS == 0:
            training_state = {
                "model_state_dict": model.state_dict(),
                "optimizer_state_dict": optimizer.state_dict(),
                "train_step": step,
            }
            torch.save(training_state, state_path)

        # Save model snapshot every KEEP_EVERY_STEPS
        if step % KEEP_EVERY_STEPS == 0 and step > 0:
            torch.save(
                model.state_dict(),
                os.path.join(OUT_DIR, f"model_{step}.pt"),
            )    
```

```python
# code to actually train the model
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = TransformerModel(
    n_dims = N_DIMS,
    n_positions = N_POSITIONS,
    n_embd = N_EMBD,
    n_layer = N_ATT_LAYERS,
    n_head = N_ATT_HEADS,
)
model.to(device)
train(model)
```

```python
def load_model(state_path, device=None):
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    model = TransformerModel(
        n_dims = N_DIMS,
        n_positions = N_POSITIONS,
        n_embd = N_EMBD,
        n_layer = N_ATT_LAYERS,
        n_head = N_ATT_HEADS,
    ).to(device)

    state = torch.load(state_path, map_location = device, weights_only = True)
    state_dict = state["model_state_dict"]
    
    _, _ = model.load_state_dict(state_dict, strict = False)

    model.eval()
    return model
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
def least_squares_baseline(samples, n_points, n_dims, sparsity = None):
    lstsq_losses = []
    
    for _ in tqdm(range(10000)):
        xs = sample_gaussian_xs(
            batch_size = 1,
            n_points = n_points,
            n_dims = n_dims,
            n_dims_trunc = n_dims,
            device = device,
        )[0]
        w = sample_linear_regression_weights(
            batch_size = 1,
            n_dims = n_dims,
            device = device,
            sparsity_s = sparsity
        )
        ys = evaluate_linear_regression(xs, w)[0]
        
        lstsq_mses = []
        for i in range(len(xs)):
            X, y = xs[:i], ys[:i]
            x_query = xs[i].cpu().numpy()
            y_query = ys[i].cpu().numpy()
            
            w_hat = np.linalg.lstsq(X, y, rcond = None)[0]
            y_pred_lstsq = x_query @ w_hat
            lstsq_mse = (y_pred_lstsq - y_query) ** 2

            if sparsity:
                lstsq_mse /= sparsity
            else:
                lstsq_mse /= n_dims
                
            lstsq_mses.append(lstsq_mse)
        
        lstsq_losses.append(lstsq_mses)
    
    lstsq_loss = np.array(lstsq_losses).mean(axis = 0)
    return lstsq_loss

lstsq_loss = least_squares_baseline(10000, 41, 20)
plt.plot(np.arange(len(lstsq_loss)), lstsq_loss)
plt.xlabel("in-context examples")
plt.ylabel("squared error")
plt.xlim(0, 40)
plt.ylim(0, 1.1)
plt.grid(True)
plt.show()
```

```python
@torch.no_grad()
def eval_model(model, n_points, n_dims, num_batches, batch_size, sparsity):
    device = next(model.parameters()).device
    model.eval()

    sum_sq = None 
    n_total = 0

    for _ in tqdm(range(num_batches)):
        xs = sample_gaussian_xs(
            batch_size = batch_size,
            n_points = n_points,
            n_dims = n_dims,
            n_dims_trunc = n_dims,
            device = device,
        )
        w = sample_linear_regression_weights(
            batch_size = batch_size,
            n_dims = n_dims,
            device = device,
            sparsity_s = sparsity
        )
        ys = evaluate_linear_regression(xs, w)

        preds = model(xs, ys)
        sq_err = (preds - ys) ** 2
        if sparsity:
            sq_err /= sparsity
        else:
            sq_err /= n_dims

        batch_sum = sq_err.sum(dim = 0)
        sum_sq = sum_sq + batch_sum if sum_sq is not None else batch_sum

        n_total += sq_err.shape[0]

    mse_curve = (sum_sq / n_total).cpu().numpy()
    ks = np.arange(n_points)

    return ks, mse_curve
```

```python
model = load_model("models/linear_regression/state.pt")

ks, mses = eval_model(
    model,
    n_points = CURR_POINTS_END,
    n_dims = CURR_DIMS_END,
    num_batches = 157,
    batch_size = BATCH_SIZE,
    sparsity = None
)

plt.figure(figsize = (10, 6))
plt.plot(np.arange(len(lstsq_loss)), lstsq_loss, label = "Least Squares", color = "orange")
plt.plot(ks, mses, label = "Transformer")
plt.xlabel("in-context examples")
plt.ylabel("squared error")
plt.ylim(0.0, max(1.2, mses.max() * 1.05))
plt.xlim(0, CURR_POINTS_END - 1)
plt.grid(alpha = 0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
model = load_model("models/sparse_linear_regression/state.pt")

ks, mses = eval_model(
    model,
    n_points = CURR_POINTS_END,
    n_dims = CURR_DIMS_END,
    num_batches = 157,
    batch_size = BATCH_SIZE,
    sparsity = SPARSITY
)

plt.figure(figsize = (10, 6))
plt.plot(ks, mses, label = "Transformer")
plt.xlabel("in-context examples")
plt.ylabel("squared error")
plt.ylim(0.0, max(1.2, mses.max() * 1.05))
plt.xlim(0, CURR_POINTS_END - 1)
plt.grid(alpha = 0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
from sklearn.linear_model import Lasso

lasso_alpha = 0.1  # hyperparameter, you can tune this
lasso_losses = []

for _ in tqdm(range(10000)):
    xs = sample_gaussian_xs(
        batch_size = 1,
        n_points = CURR_POINTS_END,
        n_dims = N_DIMS,
        n_dims_trunc = CURR_DIMS_END,
        device = device,
    )[0]  # (T, d)
    w = sample_linear_regression_weights(
        batch_size = 1,
        n_dims = N_DIMS,
        device = device,
        sparsity_s = SPARSITY
    )
    ys = evaluate_linear_regression(xs, w)[0]  # (T, 1) or (T,)

    lasso_mses = []

    # start from 1 so we have at least 1 training point
    for i in range(1, len(xs)):
        # shared inputs
        X = xs[:i]          # (i, d)
        y = ys[:i]          # (i, 1) or (i,)
        x_query = xs[i].cpu().numpy()
        y_query = ys[i].cpu().numpy()

        # convert to numpy for sklearn
        X_np = X.cpu().numpy()
        y_np = y.cpu().numpy().reshape(-1)  # flatten to shape (i,)

        # LASSO fit (no intercept to match linear-regression setup)
        lasso = Lasso(
            alpha = lasso_alpha,
            fit_intercept = False,
            max_iter = 10_000
        )
        lasso.fit(X_np, y_np)

        # prediction on query point
        y_pred_lasso = lasso.predict(x_query[None, :])[0]

        # normalized squared error (divide by d = N_DIMS)
        lasso_mse = (y_pred_lasso - y_query) ** 2 / SPARSITY
        lasso_mses.append(float(lasso_mse))

    lasso_losses.append(lasso_mses)

lasso_loss = np.array(lasso_losses).mean(axis = 0)

lstsq_loss = least_squares_baseline(10000, 41, 20, 3)
plt.plot(np.arange(len(lstsq_loss)), lstsq_loss)
plt.plot(np.arange(1, len(lasso_loss) + 1), lasso_loss, label = "LASSO")
plt.xlabel("in-context examples")
plt.ylabel("squared error")
plt.xlim(0, 40)
plt.ylim(0, 1.1)
plt.grid(True)
plt.legend()
plt.show()
```

```python
plt.figure(figsize = (10, 6))
plt.plot(ks, mses, label = "Transformer")
plt.xlabel("in-context examples")
plt.ylabel("squared error")
plt.ylim(0.0, max(1.2, mses.max() * 1.05))
plt.plot(np.arange(len(lstsq_loss)), lstsq_loss, label = "Least Squares", color = "orange")
plt.plot(np.arange(1, len(lasso_loss) + 1), lasso_loss, label = "LASSO", color = "red")
plt.xlabel("in-context examples")
plt.ylabel("squared error")
plt.xlim(0, 40)
plt.ylim(0, 1.1)
plt.grid(True)
plt.legend()
plt.show()
```
