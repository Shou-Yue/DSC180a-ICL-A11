```python
import torch
import torch.nn as nn
from transformers import GPT2Model, GPT2Config
from tqdm import tqdm
from types import SimpleNamespace
```

```python
if torch.backends.mps.is_available():
    device = torch.device("mps")
    print("Device: Apple Silicon")
elif torch.cuda.is_available():
    device = torch.device("cuda")
    print("Device: GPU")
else:
    device = torch.device("cpu")
    print("Device: CPU")
```

```python
# config parameters
d = 20 # dimension of input samples
n = 20 # number of input samples
s = None # nonzero weights in ground truth weight vector, None if not doing sparse linear regression
```

```python
def generate_batch(batch_size, n, curr_d, max_d, device, s = None):
    # 1. Sample all x's in the current active subspace, zero-pad to max_d.
    # xs[:, :, :curr_d] ~ N(0, 1), xs[:, :, curr_d:] = 0
    xs = torch.zeros(batch_size, n, max_d, device = device)
    xs[:, :, :curr_d] = torch.randn(batch_size, n, curr_d, device = device)

    # 2. Sample ground-truth weights w in the same subspace, zero-pad to max_d.
    ws = torch.zeros(batch_size, max_d, device = device)

    if s is None or s >= curr_d:
        # Dense case: all first curr_d coords are active
        ws[:, :curr_d] = torch.randn(batch_size, curr_d, device = device)
    else:
        # Sparse case: exactly s nonzero coords within the first curr_d dims
        # for each task in the batch.
        for b in range(batch_size):
            # Choose s random active coordinates inside [0, curr_d)
            idx = torch.randperm(curr_d, device = device)[:s]
            ws[b, idx] = torch.randn(s, device = device)

    # 3. Evaluate y = x @ w for each task and point.
    # xs: (B, n, max_d), ws: (B, max_d) -> ys: (B, n)
    ys = torch.einsum("bnd,bd->bn", xs, ws)

    return xs, ys
```

```python
class Curriculum:
    def __init__(self, dims_start: int, dims_end: int, dims_inc: int, 
                 points_start: int, points_end: int, points_inc: int, interval: int):
        self.n_dims = dims_start
        self.dims_end = dims_end
        self.dims_inc = dims_inc

        self.n_points = points_start
        self.points_end = points_end
        self.points_inc = points_inc

        self.interval = interval
        self.step_count = 0

    def update(self):
        self.step_count += 1

        if self.step_count % self._interval == 0:
            self.n_dims = min(self.n_dims + self.dims_inc, self.dims_end)
            self.n_points = min(self.n_points + self.points_inc, self.points_end)
```

```python
class TransformerModel(nn.Module):
    def __init__(self, n_dims, n_positions, n_embd = 128, n_layer = 1, n_head = 1):
        super(TransformerModel, self).__init__()
        configuration = GPT2Config(
            n_positions = 2 * n_positions,
            n_embd = n_embd,
            n_layer = n_layer,
            n_head = n_head,
            resid_pdrop = 0.0,
            embd_pdrop = 0.0,
            attn_pdrop = 0.0,
            use_cache = False,
        )
        self.name = f"gpt2_embd={n_embd}_layer={n_layer}_head={n_head}"

        self.n_positions = n_positions
        self.n_dims = n_dims
        self._read_in = nn.Linear(n_dims, n_embd)
        self._backbone = GPT2Model(configuration)
        self._read_out = nn.Linear(n_embd, 1)

    @staticmethod
    def _combine(xs_b, ys_b):
        """Interleaves the x's and the y's into a single sequence."""
        bsize, points, dim = xs_b.shape
        ys_b_wide = torch.cat(
            (
                ys_b.view(bsize, points, 1),
                torch.zeros(bsize, points, dim - 1, device = ys_b.device),
            ),
            axis = 2,
        )
        zs = torch.stack((xs_b, ys_b_wide), dim = 2)
        zs = zs.view(bsize, 2 * points, dim)
        return zs

    def forward(self, xs, ys, inds = None):
        if inds is None:
            inds = torch.arange(ys.shape[1], device = ys.device)
        else:
            inds = torch.tensor(inds, device = ys.device)
            if max(inds) >= ys.shape[1] or min(inds) < 0:
                raise ValueError("inds contain indices where xs and ys are not defined")
        zs = self._combine(xs, ys)
        embeds = self._read_in(zs)
        output = self._backbone(inputs_embeds = embeds).last_hidden_state
        prediction = self._read_out(output)
        # prediction has shape (B, 2 * n_points, 1); take only x positions
        return prediction[:, ::2, 0][:, inds]  # predict only on xs
```

```python
class Curriculum:
    def __init__(self, args):
        """
        args.dims and args.points each contain: start, end, inc, interval
        - 'inc' = increment value
        - 'interval' = how many steps before applying the increment
        """
        self.n_dims = args.dims.start
        self.n_points = args.points.start
        self.n_dims_schedule = args.dims
        self.n_points_schedule = args.points
        self.step_count = 0

    def update(self):
        self.step_count += 1
        self.n_dims = self._update_var(self.n_dims, self.n_dims_schedule)
        self.n_points = self._update_var(self.n_points, self.n_points_schedule)

    def _update_var(self, var, schedule):
        if self.step_count % schedule.interval == 0:
            var += schedule.inc
        return min(var, schedule.end)
```

```python
def train_step(model, xs, ys, optimizer, loss_fn):
    optimizer.zero_grad()
    preds = model(xs, ys)  # (B, n_points)
    loss = loss_fn(preds, ys)
    loss.backward()
    optimizer.step()
    return loss.item(), preds.detach()
```

```python
max_n_dims = 20      # fixed input dimension for the model
max_n_points = 20    # max number of (x, y) pairs

model = TransformerModel(
    n_dims = max_n_dims,
    n_positions = max_n_points,
    n_embd = 128,
    n_layer = 1,       # single attention layer
    n_head = 1,        # single head
).to(device)
model.train()

# ------------------------------------------------------------------
# Curriculum: from d=2, n=2 -> d=20, n=20
# ------------------------------------------------------------------
steps_per_stage = 1000  # how long to stay at each (d, n)

curriculum_cfg = SimpleNamespace(
    dims = SimpleNamespace(start = 2, end = 20, inc = 1, interval = steps_per_stage),
    points = SimpleNamespace(start = 2, end = 20, inc = 1, interval = steps_per_stage),
)
curriculum = Curriculum(curriculum_cfg)

# ------------------------------------------------------------------
# Training hyperparameters
# ------------------------------------------------------------------
batch_size = 64
num_steps = 20_000
learning_rate = 1e-4

optimizer = torch.optim.Adam(model.parameters(), lr = learning_rate)
loss_fn = nn.MSELoss()

pbar = tqdm(range(num_steps))

for step in pbar:
    n_points = curriculum.n_points
    n_dims = curriculum.n_dims

    # Ensure within model capacity
    assert n_points <= max_n_points
    assert n_dims <= max_n_dims

    xs, ys = generate_batch(
        batch_size = batch_size,
        n_points = n_points,
        n_dims_active = n_dims,
        n_dims_full = max_n_dims,
        device = device,
    )

    loss, preds = train_step(model, xs, ys, optimizer, loss_fn)
    curriculum.update()

    pbar.set_description(f"loss {loss:.6f} | d={n_dims} n={n_points}")

torch.save(model.state_dict(), "transformer_curriculum_mps.pt")
print("Training complete. Saved to transformer_curriculum_mps.pt")
```

```python
import torch
from torch.nn.functional import mse_loss

device = "mps" if torch.backends.mps.is_available() else "cpu"
model = TransformerModel(n_dims=20, n_positions=20, n_embd=128, n_layer=1, n_head=1).to(device)
model.load_state_dict(torch.load("transformer_curriculum_mps.pt", map_location=device, weights_only=True))
model.eval()

def eval_generalization(model, n_dims, n_points, device="mps", trials=100):
    losses = []
    for _ in range(trials):
        xs = torch.randn(1, n_points, n_dims, device=device)
        w = torch.randn(1, n_dims, device=device)
        ys = torch.einsum("btd,bd->bt", xs, w)

        # Hold out last point as query
        context_x, query_x = xs[:, :-1, :], xs[:, -1:, :]
        context_y, query_y = ys[:, :-1], ys[:, -1:]

        # Pad context so shapes match model.n_dims
        pad = model.n_dims - n_dims
        if pad > 0:
            context_x = torch.cat([context_x,
                                   torch.zeros(1, n_points - 1, pad, device=device)], dim=-1)
            query_x = torch.cat([query_x,
                                 torch.zeros(1, 1, pad, device=device)], dim=-1)

        # Combine context + query
        xs_full = torch.cat([context_x, query_x], dim=1)
        ys_full = torch.cat([context_y, torch.zeros_like(query_y)], dim=1)

        with torch.no_grad():
            preds = model(xs_full, ys_full)
        loss = mse_loss(preds[:, -1:], query_y)
        losses.append(loss.item())
    return sum(losses)/len(losses)

test_mse = eval_generalization(model, n_dims=10, n_points=10)
print(f"Test MSE on unseen linear tasks: {test_mse:.6f}")
```
