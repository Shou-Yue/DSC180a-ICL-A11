## Overparameterized Gradient Descent (d >> n)

```python
import numpy as np
import sys
```

```python
# set up dimension and number of points
d = 20
n = 10

# randomly generate weight vector, inputs, and outputs
w = np.random.normal(loc = 0.0, scale = 1.0, size = d)
X = np.random.normal(loc = 0.0, scale = 1.0, size = (n, d))
y = X @ w
```

```python
# loss function
def squared_loss(y_pred, y):
    return np.mean((y_pred - y) ** 2)
```

```python
# initial guess for w
w_hat = np.zeros(d)

# hyperparameters
lr = 0.01
iters = 5001

# gradient descent loop
for step in range(iters):
    y_pred = X @ w_hat
    loss = squared_loss(y_pred, y)
    
    grad = (2.0 / n) * X.T @ (y_pred - y)
    w_hat -= lr * grad
    
    if step % 1000 == 0:
        print(f"Iterations: {step:5d} | Loss {loss:.6f}")
```
