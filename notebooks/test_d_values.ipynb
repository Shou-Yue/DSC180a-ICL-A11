```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent   # This is DSC180a-ICL-A11
sys.path.insert(0, str(PROJECT_ROOT))

%load_ext autoreload
%autoreload 2

import matplotlib.pyplot as plt
from icl_classification.classification_icl import run_single_experiment
```

```python
ds = [10, 20, 50, 100, 200, 500, 800, 1000]
train_accs = []
val_accs = []
ic_accs = []

for d in ds:
    B, steps, label_flip_p = 100, 300, 0.2
    R = d ** 0.1
    params = (d, B, R, R, steps, label_flip_p)
    _, train_acc, val_acc, ic_acc = run_single_experiment(params=params, base_results_dir="test_results/", use_cuda=False)
    train_accs.append(train_acc)
    val_accs.append(val_acc)
    ic_accs.append(ic_acc)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(ds, train_accs, marker='o', label="Train Acc")
plt.plot(ds, val_accs, marker='o', label="Val Acc")
plt.plot(ds, ic_accs, marker='o', label="In-context Acc")

plt.xlabel("Dimension d")
plt.ylabel("Accuracy")
plt.title("Accuracy vs Input Dimension: R = d^0.1")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```
