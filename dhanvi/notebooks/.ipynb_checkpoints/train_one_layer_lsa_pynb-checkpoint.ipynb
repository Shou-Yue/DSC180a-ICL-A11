```python
import math, random
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt

device = "cuda" if torch.cuda.is_available() else "cpu"
torch.manual_seed(0); random.seed(0)
```

```python
class LinearTaskSampler:
    def __init__(self, d=10, n=10, batch_tasks=2048, out_dim=1, x_range=1.0):
        self.d, self.n, self.B = d, n, batch_tasks
        self.out_dim = out_dim
        self.x_range = x_range

    def sample_batch(self):
        # Teacher W: [B, out_dim, d]
        W = torch.randn(self.B, self.out_dim, self.d, device=device)
        # Context X: [B, n, d], U(-x_range, x_range)
        X = (2*self.x_range)*torch.rand(self.B, self.n, self.d, device=device) - self.x_range
        # Y = X @ W^T  -> [B, n, out_dim]
        Y = torch.matmul(X, W.transpose(1,2))  # [B,n,1]
        # Query x_test, y_test
        xq = (2*self.x_range)*torch.rand(self.B, self.d, device=device) - self.x_range     # [B, d]
        yq = torch.matmul(xq.unsqueeze(1), W.transpose(1,2)).squeeze(1)                   # [B, 1]
        return X, Y, xq, yq, W
```

```python
# =========================
# Linear Self-Attention (1 head, no softmax/LN)
# =========================
class LinearSelfAttention1L(nn.Module):
    """
    Single-head linear self-attention:
      e_j <- e_j + P * V * K^T * (Q e_j)
    Tokens are concatenated [x, y], dim = d + out_dim.
    """
    def __init__(self, token_dim):
        super().__init__()
        D = token_dim
        self.W_Q = nn.Linear(D, D, bias=False)
        self.W_K = nn.Linear(D, D, bias=False)
        self.W_V = nn.Linear(D, D, bias=False)
        self.P    = nn.Linear(D, D, bias=False)

        # tiny init helps stability, like the paper
        scale = 2e-3
        for m in [self.W_Q, self.W_K, self.W_V, self.P]:
            nn.init.trunc_normal_(m.weight, std=scale, a=-3*scale, b=3*scale)

    def forward(self, tokens):
        """
        tokens: [B, T, D]
        returns: [B, T, D] updated
        """
        Q = self.W_Q(tokens)          # [B, T, D]
        K = self.W_K(tokens)          # [B, T, D]
        V = self.W_V(tokens)          # [B, T, D]
        # KQT[i,j] = <K_i, Q_j>
        KQT = torch.einsum("bid,bjd->bij", K, Q)         # [B, T, T]
        att = torch.einsum("bij,bid->bjd", KQT, V)       # sum_i KQT[i,j] * V[i]
        out = tokens + self.P(att)
        return out
```

```python
# =========================
# Token utilities
# =========================
def build_tokens(X, Y, xq, yq_init=None):
    """
    Concatenate [x, y] per context token, and [xq, yq_init] for query.
    yq_init = 0 by default (equivalent to W0 = 0 in the paper).
    shapes:
      X: [B, n, d], Y: [B, n, 1], xq: [B, d], yq_init: [B, 1]
    """
    B, n, d = X.shape
    out_dim = Y.shape[-1]
    if yq_init is None:
        yq_init = torch.zeros(B, out_dim, device=X.device)
    ctx = torch.cat([X, Y], dim=-1)                      # [B, n, d+out_dim]
    qry = torch.cat([xq, yq_init], dim=-1).unsqueeze(1)  # [B, 1, d+out_dim]
    tokens = torch.cat([ctx, qry], dim=1)                # [B, n+1, d+out_dim]
    return tokens

def readout_y_pred(updated_tokens, out_dim=1):
    # last token is the query; take its last out_dim entries as prediction
    return updated_tokens[:, -1, -out_dim:]             # [B, out_dim]
```

```python
# =========================
# GD baseline (1 step from W0 = 0) + optimal eta*
# =========================
def gd_one_step_pred(X, Y, xq, lr):
    """
    One GD step for y = w^T x from w0=0 (scalar output).
    Delta w = (lr/N) * sum_i y_i x_i
    yhat = (Delta w)^T xq
    """
    B, n, d = X.shape
    # sum_i y_i x_i -> [B,d] using broadcasting
    sum_yx = (X * Y) .sum(dim=1)           # [B,d]
    dw = (lr / n) * sum_yx                 # [B,d]
    yhat = (dw * xq).sum(dim=-1, keepdim=True)  # [B,1]
    return yhat

def gd_eta_star_batch(X, Y, xq, yq):
    """
    MSE-optimal eta* for 1-step GD from W0=0 (per batch).
    GD prediction = eta * s, where s = (1/n) xq^T sum_i y_i x_i
    eta* = sum_b s_b y_b / sum_b s_b^2
    """
    B, n, d = X.shape
    sum_yx = (X * Y).sum(dim=1)                     # [B,d]
    s = (xq * sum_yx).sum(dim=-1) / n               # [B]
    num = (s * yq.squeeze(-1)).sum()
    den = (s * s).sum().clamp_min(1e-12)
    return (num / den).item()

# =========================
# Training
# =========================
def train_lsa(
    d=10, n=10, batch_tasks=8192, steps=3000,
    x_range=1.0, lr_model=2e-3, report_every=200,
    distill_frac=0.7   # proportion of steps to distill to 1-step GD
):
    sampler = LinearTaskSampler(d=d, n=n, batch_tasks=batch_tasks, out_dim=1, x_range=x_range)
    token_dim = d + 1
    lsa = LinearSelfAttention1L(token_dim).to(device)

    opt = torch.optim.Adam(lsa.parameters(), lr=lr_model)
    hist = {"step": [], "lsa_eval": [], "gd_eval": [], "sens_cos": [], "pred_L2": []}

    switch_step = int(distill_frac * steps)

    for step in range(1, steps+1):
        # ----- sample tasks -----
        X, Y, xq, yq, _ = sampler.sample_batch()
        tokens = build_tokens(X, Y, xq, yq_init=None)
        pred_lsa = readout_y_pred(lsa(tokens), out_dim=1)

        # ----- curriculum: distill -> truth -----
        if step <= switch_step:
            # compute eta* and 1-step GD teacher on THIS batch
            with torch.no_grad():
                eta_star = gd_eta_star_batch(X, Y, xq, yq)
                teacher = gd_one_step_pred(X, Y, xq, lr=eta_star)
            loss = F.mse_loss(pred_lsa, teacher)
        else:
            loss = F.mse_loss(pred_lsa, yq)

        opt.zero_grad()
        loss.backward()
        nn.utils.clip_grad_norm_(lsa.parameters(), max_norm=10.0)
        opt.step()

        # ----- periodic eval -----
        if step % report_every == 0 or step == 1:
            lsa.eval()
            with torch.no_grad():
                Xv, Yv, xqv, yqv, _ = sampler.sample_batch()
                eta_star_v = gd_eta_star_batch(Xv, Yv, xqv, yqv)
                gd_pred = gd_one_step_pred(Xv, Yv, xqv, lr=eta_star_v)
                gd_loss = F.mse_loss(gd_pred, yqv)

                lsa_pred = readout_y_pred(lsa(build_tokens(Xv, Yv, xqv, None)))
                lsa_loss = F.mse_loss(lsa_pred, yqv)

            # alignment (fresh batch, keep grads on xq only)
            Xv2, Yv2, xqv2, yqv2, _ = sampler.sample_batch()
            xq_req = xqv2.clone().detach().requires_grad_(True)
            sens_lsa = torch.autograd.grad(
                readout_y_pred(lsa(build_tokens(Xv2, Yv2, xq_req, None))).sum(), xq_req
            )[0]

            xq_req2 = xqv2.clone().detach().requires_grad_(True)
            eta_star2 = gd_eta_star_batch(Xv2, Yv2, xqv2, yqv2)
            sens_gd = torch.autograd.grad(
                gd_one_step_pred(Xv2, Yv2, xq_req2, lr=eta_star2).sum(), xq_req2
            )[0]

            cos = F.cosine_similarity(sens_lsa, sens_gd, dim=-1).mean().item()
            with torch.no_grad():
                pred_L2 = F.mse_loss(
                    readout_y_pred(lsa(build_tokens(Xv2, Yv2, xqv2, None))),
                    gd_one_step_pred(Xv2, Yv2, xqv2, lr=eta_star2)
                ).sqrt().item()

            hist["step"].append(step)
            hist["lsa_eval"].append(lsa_loss.item())
            hist["gd_eval"].append(gd_loss.item())
            hist["sens_cos"].append(cos)
            hist["pred_L2"].append(pred_L2)

            phase = "DISTILL" if step <= switch_step else "TRUTH"
            print(f"step {step:4d} [{phase}] | LSA eval {lsa_loss.item():.4f} | GD eval {gd_loss.item():.4f} "
                  f"| sens cos {cos:.3f} | pred L2 {pred_L2:.4f} | eta* {eta_star_v:.4f}")
            lsa.train()

    return lsa, hist


# =========================
# OOD evaluation + rollout utils
# =========================
def eval_ood_curve(lsa, d=10, n=10, alphas=(0.5,1.0,1.5,2.0,3.0), B=4096):
    lsa.eval(); res=[]
    for a in alphas:
        sampler = LinearTaskSampler(d=d, n=n, batch_tasks=B, out_dim=1, x_range=a)
        with torch.no_grad():
            X, Y, xq, yq, _ = sampler.sample_batch()
            eta = gd_eta_star_batch(X, Y, xq, yq)
            gd = gd_one_step_pred(X, Y, xq, lr=eta)
            gd_loss = F.mse_loss(gd, yq).item()
            lsa_pred = readout_y_pred(lsa(build_tokens(X, Y, xq, None)))
            lsa_loss = F.mse_loss(lsa_pred, yq).item()
        res.append((a, lsa_loss, gd_loss))
    return res

def rollout_k_steps(lsa, X, Y, xq, steps=2, lam=0.75):
    """
    Apply the *same trained LSA layer* repeatedly (post-train), with residual dampening lam.
    Returns yhat after 'steps' applications.
    """
    lsa.eval()
    with torch.no_grad():
        toks = build_tokens(X, Y, xq, None)
        cur = toks
        for _ in range(steps):
            upd = lsa(cur)
            cur = cur + lam * (upd - cur)
        yhat = readout_y_pred(cur)
    return yhat

# =========================
# Run training
# =========================
d, n = 10, 10
lsa, hist = train_lsa(d=d, n=n, steps=2000, batch_tasks=2048, x_range=1.0, lr_model=1e-3, report_every=200)
```

```python
# =========================
# Plots: Loss vs step, Alignment, Prediction gap
# =========================
plt.figure(figsize=(6,4))
plt.plot(hist["step"], hist["lsa_eval"], label="LSA eval")
plt.plot(hist["step"], hist["gd_eval"], label="1-step GD eval")
plt.xlabel("Step"); plt.ylabel("MSE"); plt.title("LSA vs 1-step GD (eval)"); plt.legend(); plt.grid(True); plt.show()

plt.figure(figsize=(6,4))
plt.plot(hist["step"], hist["pred_L2"])
plt.xlabel("Step"); plt.ylabel("RMSE gap"); plt.title("Prediction gap: LSA vs GD"); plt.grid(True); plt.show()

# =========================
# OOD curve (input scaling)
# =========================
ood = eval_ood_curve(lsa, d=d, n=n, alphas=(0.5,1.0,1.5,2.0,3.0))
alphas = [r[0] for r in ood]; lsa_l = [r[1] for r in ood]; gd_l = [r[2] for r in ood]
plt.figure(figsize=(6,4))
plt.plot(alphas, lsa_l, marker="o", label="LSA")
plt.plot(alphas, gd_l, marker="o", label="1-step GD")
plt.xlabel("α (input scale)"); plt.ylabel("MSE"); plt.title("OOD scaling"); plt.legend(); plt.grid(True); plt.show()

# =========================
# (Optional) Quick rollout check: 2 steps vs 2-step GD
# =========================
sampler = LinearTaskSampler(d=d, n=n, batch_tasks=4096, out_dim=1, x_range=1.0)
with torch.no_grad():
    Xr, Yr, xqr, yqr, _ = sampler.sample_batch()
    # rollout twice
    yhat_roll = rollout_k_steps(lsa, Xr, Yr, xqr, steps=2, lam=0.75)
    # two-step GD (reuse eta* for simplicity)
    eta1 = gd_eta_star_batch(Xr, Yr, xqr, yqr)
    # step 1
    sum_yx = (Xr * Yr).sum(dim=1)             # [B,d]
    dw1 = (eta1 / n) * sum_yx                 # [B,d]
    yhat_ctx1 = torch.bmm(Xr, dw1.unsqueeze(-1)).squeeze(-1).unsqueeze(-1)  # [B,n,1]
    R = Yr - yhat_ctx1                         # residuals [B,n,1]
    # step 2
    sum_Rx = (Xr * R).sum(dim=1)              # [B,d]
    dw2 = (eta1 / n) * sum_Rx
    w2 = dw1 + dw2
    yhat_gd2 = (w2 * xqr).sum(dim=-1, keepdim=True)  # [B,1]
    print("Rollout(2) MSE:", F.mse_loss(yhat_roll, yqr).item(), "| 2-step GD MSE:", F.mse_loss(yhat_gd2, yqr).item())
```

```python
plt.style.use("seaborn-v0_8-whitegrid")
plt.figure(figsize=(5,3))
plt.plot(hist["step"], hist["lsa_eval"], lw=2, label="LSA (learned)")
plt.plot(hist["step"], hist["gd_eval"], lw=2, linestyle="--", label="1-step GD (teacher)")
plt.xlabel("Training steps")
plt.ylabel("MSE")
plt.title("Loss alignment: LSA reproduces 1-step GD")
plt.legend(frameon=False)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(5,3))
plt.plot(hist["step"], hist["pred_L2"], lw=2, color="tab:blue")
plt.xlabel("Training steps")
plt.ylabel("RMSE gap")
plt.title("Prediction agreement between LSA and GD")
plt.tight_layout()
plt.show()
```

```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt

device = next(lsa.parameters()).device  # use same device as your trained layer

# --- minimal data sampler (same distribution as training) ---
def sample_batch(d=10, n=10, batch=1024, x_range=1.0, device=device):
    w = torch.randn(batch, d, device=device)                       # teacher weights
    X = (torch.rand(batch, n, d, device=device)*2 - 1)*x_range     # (B, n, d)
    Y = (X * w.unsqueeze(1)).sum(-1, keepdim=True)                 # (B, n, 1)
    xq = (torch.rand(batch, d, device=device)*2 - 1)*x_range       # (B, d)
    y_true = (xq * w).sum(-1, keepdim=True)                        # (B, 1)
    return X, Y, xq, y_true

@torch.no_grad()
def test_lsa_simple(lsa, d=10, n=10, batch=1024, x_range=1.0):
    lsa.eval()
    X, Y, xq, y_true = sample_batch(d, n, batch, x_range, device)
    tokens = build_tokens(X, Y, xq, None)
    y_pred = readout_y_pred(lsa(tokens))

    mse = F.mse_loss(y_pred, y_true).item()
    print(f"MSE(LSA vs truth): {mse:.4f}  (batch={batch}, d={d}, n={n})")

    # quick visual: y_true vs y_pred
    yt = y_true.detach().cpu().numpy().ravel()
    yp = y_pred.detach().cpu().numpy().ravel()
    lo, hi = float(min(yt.min(), yp.min())), float(max(yt.max(), yp.max()))

    plt.figure(figsize=(5,5))
    plt.scatter(yt, yp, s=6, alpha=0.6)
    plt.plot([lo, hi], [lo, hi])  # y=x line
    plt.xlabel("y_true")
    plt.ylabel("y_pred (LSA)")
    plt.title("y_true vs y_pred")
    plt.grid(True)
    plt.show()

# --- run the simple test ---
test_lsa_simple(lsa, d=10, n=10, batch=2048, x_range=1.0)
```
